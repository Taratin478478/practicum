```python
import subprocess
import sys

def check_current_cuda():
    print("=== Checking Current CUDA Setup ===")
    
    # Check nvidia-smi (driver CUDA)
    try:
        result = subprocess.run(['nvidia-smi'], capture_output=True, text=True, shell=True)
        for line in result.stdout.split('\n'):
            if 'CUDA Version' in line:
                print(f"Driver CUDA: {line.strip()}")
    except:
        print("✗ nvidia-smi not available")
    
    # Check nvcc (toolkit CUDA)
    try:
        result = subprocess.run(['nvcc', '--version'], capture_output=True, text=True)
        for line in result.stdout.split('\n'):
            if 'release' in line:
                print(f"Toolkit CUDA: {line.strip()}")
    except:
        print("✗ nvcc not available")
    
    # Check conda CUDA
    try:
        result = subprocess.run(['conda', 'list', 'cudatoolkit'], capture_output=True, text=True)
        if 'cudatoolkit' in result.stdout:
            print("Conda CUDA toolkit installed")
            for line in result.stdout.split('\n'):
                if 'cudatoolkit' in line and not line.startswith('#'):
                    print(f"Conda: {line}")
    except:
        pass

check_current_cuda()
```

```python
import tensorflow as tf
import os

# Force TensorFlow to see the conda-installed CUDA
os.environ['CUDA_VISIBLE_DEVICES'] = '0'

print("=== Testing TensorFlow GPU ===")
print(f"TensorFlow version: {tf.__version__}")

# List physical devices
gpus = tf.config.list_physical_devices('GPU')
print(f"GPU devices: {gpus}")

if gpus:
    # Set memory growth to avoid allocating all GPU memory
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)
    
    # Test computation on GPU
    with tf.device('/GPU:0'):
        a = tf.constant([[1.0, 2.0], [3.0, 4.0]])
        b = tf.constant([[1.0, 1.0], [0.0, 1.0]])
        c = tf.matmul(a, b)
        
        print("✓ GPU computation successful!")
        print(f"Device: {c.device}")
        print(f"Result: {c.numpy()}")
else:
    print("✗ No GPU devices detected")
```
